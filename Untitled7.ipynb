```python
# ============================================================
# 🌱 ECOVISION - COMPLETE ONE CELL VERSION
# ============================================================

# -------------------- INSTALL LIBRARIES ---------------------

import sys
import subprocess

packages = [
    "gradio",
    "pandas",
    "numpy",
    "pillow"
]

for package in packages:
    subprocess.check_call([
        sys.executable,
        "-m",
        "pip",
        "install",
        "-q",
        package
    ])

print("✅ All required libraries installed!")


# -------------------- IMPORTS -------------------------------

import gradio as gr
import pandas as pd
from datetime import datetime


# ============================================================
# 🌱 ECOVISION DATA
# ============================================================

WASTE_DATA = {

    "Plastic": {
        "action": "♻️ Recycle",
        "points": 5,
        "ideas": [
            "Reuse plastic containers as storage boxes.",
            "Use plastic bottles as plant holders.",
            "Send clean plastic to a recycling center."
        ]
    },

    "Paper": {
        "action": "♻️ Recycle",
        "points": 5,
        "ideas": [
            "Use one-sided paper for notes.",
            "Make paper crafts.",
            "Separate clean paper for recycling."
        ]
    },

    "Cardboard": {
        "action": "🔄 Reuse / Recycle",
        "points": 10,
        "ideas": [
            "Make storage organizers.",
            "Create DIY decorations.",
            "Flatten and recycle cardboard."
        ]
    },

    "Glass": {
        "action": "🔄 Reuse / Recycle",
        "points": 10,
        "ideas": [
            "Reuse glass jars as storage containers.",
            "Use jars as plant holders.",
            "Recycle glass safely."
        ]
    },

    "Metal": {
        "action": "♻️ Recycle",
        "points": 5,
        "ideas": [
            "Use cans as pen holders.",
            "Create small planters.",
            "Send metal items for recycling."
        ]
    },

    "Textile": {
        "action": "🔄 Reuse / Donate",
        "points": 10,
        "ideas": [
            "Turn old clothes into cleaning cloths.",
            "Make reusable bags.",
            "Donate wearable clothes."
        ]
    },

    "Organic": {
        "action": "🌱 Compost",
        "points": 10,
        "ideas": [
            "Use food waste for compost.",
            "Create a small home compost bin.",
            "Use compost for plants."
        ]
    },

    "E-Waste": {
        "action": "⚡ Send to an e-waste collection center",
        "points": 10,
        "ideas": [
            "Do not put electronic waste in normal bins.",
            "Give old electronics to an authorized collection center.",
            "Reuse working components when possible."
        ]
    }
}


# ============================================================
# 👤 USER ECO DATA
# ============================================================

user = {
    "points": 0,
    "earned": 0,
    "redeemed": 0,
    "reused": 0,
    "recycled": 0,
    "disposed": 0,
    "activities": [],
    "rewards": [],
    "challenges": []
}


# ============================================================
# 🛍️ REWARD STORE
# ============================================================

REWARDS = {
    "🌱 Mint Seeds": 50,
    "🌿 Basil Seeds": 50,
    "🌸 Flower Seeds": 75,
    "🥕 Vegetable Seeds": 100,
    "🪴 Small Plant": 150,
    "🌿 Herb Plant": 200,
    "🪴 Indoor Plant": 250,
    "🌳 Tree Sapling": 400
}


# ============================================================
# 🏆 LEVEL SYSTEM
# ============================================================

def get_level(points):

    if points >= 500:
        return "🌳 Eco Guardian"

    elif points >= 300:
        return "🌿 Eco Hero"

    elif points >= 150:
        return "🏆 Green Champion"

    elif points >= 50:
        return "🌱 Eco Explorer"

    else:
        return "🌎 Eco Starter"


# ============================================================
# 🔍 WASTE SCANNER
# ============================================================

def scan_waste(image):

    if image is None:

        return (
            "⚠️ Please upload an image.",
            "",
            "",
            "",
            ""
        )

    # Prototype category
    category = "Plastic"

    data = WASTE_DATA[category]

    prediction = f"""
# 🤖 AI Waste Analysis

### Detected Waste
**{category}**

### Status
✅ Image received successfully
"""

    action = f"""
# ♻️ Recommended Action

**{data["action"]}**
"""

    points = f"""
# 🌱 Eco Points

**+{data["points"]} Eco Points**
"""

    ideas = "# 💡 Reuse Ideas\n\n"

    for idea in data["ideas"]:
        ideas += f"• {idea}\n\n"

    status = """
# 📊 Analysis

✅ Image processed successfully.

ℹ️ This is the prototype version.
The trained waste-classification model can be connected later.
"""

    return prediction, action, points, ideas, status


# ============================================================
# 🌱 RECORD ECO ACTION
# ============================================================

def record_action(action):

    if not action:

        return "⚠️ Please select an action."

    if action == "I Reused It":

        points = 10
        action_name = "Reused"
        user["reused"] += 1

    elif action == "I Recycled It":

        points = 5
        action_name = "Recycled"
        user["recycled"] += 1

    else:

        points = 3
        action_name = "Disposed Properly"
        user["disposed"] += 1

    user["points"] += points
    user["earned"] += points

    user["activities"].append({

        "Date": datetime.now().strftime("%d-%m-%Y %H:%M"),

        "Action": action_name,

        "Points": points

    })

    return f"""
# ✅ Eco Action Recorded

🌱 Points Earned: **+{points}**

💰 Current Balance: **{user["points"]}**

🏆 Level: **{get_level(user["points"])}**

🌍 Keep helping the planet!
"""


# ============================================================
# 💰 ECO WALLET
# ============================================================

def wallet():

    summary = f"""
# 💰 Eco Wallet

### 🌱 Current Eco Points
**{user["points"]}**

### 📈 Total Points Earned
**{user["earned"]}**

### 🎁 Points Redeemed
**{user["redeemed"]}**

### 🏆 Current Level
**{get_level(user["points"])}**

---

### ♻️ Activity Summary

🔄 Reused: **{user["reused"]}**

♻️ Recycled: **{user["recycled"]}**

🗑️ Properly Disposed: **{user["disposed"]}**
"""

    if user["activities"]:

        table = pd.DataFrame(user["activities"])

    else:

        table = pd.DataFrame(
            columns=[
                "Date",
                "Action",
                "Points"
            ]
        )

    return summary, table


# ============================================================
# 🛍️ REWARD STORE DISPLAY
# ============================================================

def show_store():

    text = """
# 🛍️ Eco Reward Store

Exchange your Eco Points for plants and seeds 🌱

---

"""

    for reward, cost in REWARDS.items():

        text += f"""
### {reward}

💰 Cost: **{cost} Eco Points**

---

"""

    text += f"""
## 💰 Your Balance

**{user["points"]} Eco Points**
"""

    return text


# ============================================================
# 🎁 REDEEM REWARD
# ============================================================

def redeem_reward(reward):

    if not reward:

        return "⚠️ Please select a reward."

    cost = REWARDS[reward]

    if user["points"] < cost:

        return f"""
# ❌ Not Enough Points

🎁 Reward: **{reward}**

💰 Required: **{cost}**

💰 Available: **{user["points"]}**

♻️ Complete more eco actions to earn points.
"""

    user["points"] -= cost

    user["redeemed"] += cost

    user["rewards"].append({

        "Date": datetime.now().strftime("%d-%m-%Y %H:%M"),

        "Reward": reward,

        "Cost": cost,

        "Status": "Redeemed"

    })

    return f"""
# 🎉 Reward Redeemed Successfully!

🎁 Reward: **{reward}**

💰 Points Used: **{cost}**

🌱 Remaining Points: **{user["points"]}**

Thank you for choosing a greener lifestyle! 🌍
"""


# ============================================================
# 🎁 MY REWARDS
# ============================================================

def my_rewards():

    if not user["rewards"]:

        return pd.DataFrame(
            columns=[
                "Date",
                "Reward",
                "Cost",
                "Status"
            ]
        )

    return pd.DataFrame(user["rewards"])


# ============================================================
# 🏆 ECO CHALLENGES
# ============================================================

CHALLENGES = {

    "♻️ Recycle 5 Items": 20,

    "🌱 Reuse 3 Items": 20,

    "🪴 Plant a Seed": 20,

    "🚯 Avoid Single-Use Plastic": 20

}


def complete_challenge(challenge):

    if not challenge:

        return "⚠️ Please select a challenge."

    points = CHALLENGES[challenge]

    user["points"] += points

    user["earned"] += points

    user["challenges"].append(challenge)

    user["activities"].append({

        "Date": datetime.now().strftime("%d-%m-%Y %H:%M"),

        "Action": "Eco Challenge",

        "Points": points

    })

    return f"""
# 🏆 Challenge Completed!

### {challenge}

🌱 Points Earned: **+{points}**

💰 Current Balance: **{user["points"]}**

🏆 Level: **{get_level(user["points"])}**
"""


# ============================================================
# 🌍 ECO JOURNEY
# ============================================================

def eco_journey():

    total_actions = (
        user["reused"]
        + user["recycled"]
        + user["disposed"]
    )

    return f"""
# 🌍 My Eco Journey

## 🏆 Level

**{get_level(user["points"])}**

## 🌱 Eco Points

**{user["points"]}**

## ♻️ Total Eco Actions

**{total_actions}**

## 🔄 Items Reused

**{user["reused"]}**

## ♻️ Items Recycled

**{user["recycled"]}**

## 🗑️ Items Properly Disposed

**{user["disposed"]}**

## 🏆 Challenges Completed

**{len(user["challenges"])}**

---

# 🌱 EcoVision Cycle

### 🗑️ Waste

⬇️

### 🤖 AI Identification

⬇️

### ♻️ Reuse / Recycle

⬇️

### 🌱 Eco Points

⬇️

### 🪴 Plants & Seeds
"""


# ============================================================
# 🌱 GRADIO APP
# ============================================================

with gr.Blocks(
    title="EcoVision"
) as app:

    gr.Markdown(
        """
# 🌱 EcoVision

## AI-Powered Waste Reuse, Recycling & Eco-Reward Platform

### 🗑️ Waste → 🤖 AI → ♻️ Action → 🌱 Points → 🪴 Plants
"""
    )


    # ========================================================
    # TAB 1 - WASTE SCANNER
    # ========================================================

    with gr.Tab("🔍 Waste Scanner"):

        gr.Markdown(
            """
## 🤖 AI Waste Scanner

Upload an image of waste to analyze it.
"""
        )

        image_input = gr.Image(
            type="pil",
            label="📷 Upload Waste Image"
        )

        scan_button = gr.Button(
            "🔍 Analyze Waste",
            variant="primary"
        )

        prediction_output = gr.Markdown()

        action_output = gr.Markdown()

        points_output = gr.Markdown()

        ideas_output = gr.Markdown()

        status_output = gr.Markdown()

        scan_button.click(

            fn=scan_waste,

            inputs=image_input,

            outputs=[
                prediction_output,
                action_output,
                points_output,
                ideas_output,
                status_output
            ]
        )

        gr.Markdown("---")

        gr.Markdown(
            """
## 🌱 Record Your Eco Action
"""
        )

        action_dropdown = gr.Dropdown(

            choices=[
                "I Reused It",
                "I Recycled It",
                "I Disposed It Properly"
            ],

            label="Select your action"
        )

        action_button = gr.Button(
            "🌱 Earn Eco Points",
            variant="primary"
        )

        action_result = gr.Markdown()

        action_button.click(

            fn=record_action,

            inputs=action_dropdown,

            outputs=action_result
        )


    # ========================================================
    # TAB 2 - ECO WALLET
    # ========================================================

    with gr.Tab("💰 Eco Wallet"):

        gr.Markdown(
            """
# 💰 Eco Wallet

Track your Eco Points and sustainability activity.
"""
        )

        wallet_button = gr.Button(
            "🔄 Refresh Wallet"
        )

        wallet_output = gr.Markdown()

        activity_output = gr.Dataframe(
            interactive=False
        )

        wallet_button.click(

            fn=wallet,

            inputs=[],

            outputs=[
                wallet_output,
                activity_output
            ]
        )


    # ========================================================
    # TAB 3 - REWARD STORE
    # ========================================================

    with gr.Tab("🛍️ Reward Store"):

        store_output = gr.Markdown(
            show_store()
        )

        refresh_store_button = gr.Button(
            "🔄 Refresh Store"
        )

        refresh_store_button.click(

            fn=show_store,

            inputs=[],

            outputs=store_output
        )

        gr.Markdown("---")

        reward_dropdown = gr.Dropdown(

            choices=list(REWARDS.keys()),

            label="🌱 Select Plant or Seed"
        )

        redeem_button = gr.Button(
            "🎁 Redeem Reward",
            variant="primary"
        )

        redeem_output = gr.Markdown()

        redeem_button.click(

            fn=redeem_reward,

            inputs=reward_dropdown,

            outputs=redeem_output
        )


    # ========================================================
    # TAB 4 - ECO CHALLENGES
    # ========================================================

    with gr.Tab("🏆 Eco Challenges"):

        gr.Markdown(
            """
# 🏆 Eco Challenges

Complete challenges and earn extra Eco Points.
"""
        )

        challenge_dropdown = gr.Dropdown(

            choices=list(CHALLENGES.keys()),

            label="Select Challenge"
        )

        challenge_button = gr.Button(
            "🏆 Complete Challenge",
            variant="primary"
        )

        challenge_output = gr.Markdown()

        challenge_button.click(

            fn=complete_challenge,

            inputs=challenge_dropdown,

            outputs=challenge_output
        )


    # ========================================================
    # TAB 5 - MY REWARDS
    # ========================================================

    with gr.Tab("🎁 My Rewards"):

        gr.Markdown(
            """
# 🎁 My Rewards

View your redeemed plants and seeds.
"""
        )

        rewards_button = gr.Button(
            "🔄 View My Rewards"
        )

        rewards_output = gr.Dataframe(
            interactive=False
        )

        rewards_button.click(

            fn=my_rewards,

            inputs=[],

            outputs=rewards_output
        )


    # ========================================================
    # TAB 6 - ECO JOURNEY
    # ========================================================

    with gr.Tab("🌍 Eco Journey"):

        gr.Markdown(
            """
# 🌍 Your Sustainability Journey
"""
        )

        journey_button = gr.Button(
            "🌱 View My Eco Journey",
            variant="primary"
        )

        journey_output = gr.Markdown()

        journey_button.click(

            fn=eco_journey,

            inputs=[],

            outputs=journey_output
        )


    # ========================================================
    # FOOTER
    # ========================================================

    gr.Markdown(
        """
---

# 🌱 EcoVision

### Waste → Action → Points → Plants

**Building a greener future through Artificial Intelligence.**
"""
    )


# ============================================================
# 🚀 LAUNCH
# ============================================================

print("")
print("==========================================")
print("🌱 ECOVISION IS STARTING...")
print("==========================================")

app.launch(
    share=True,
    debug=True
)
```
